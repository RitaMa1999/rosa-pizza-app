# **Part I. Describe: How bad is it, really?**
#### **a. Function about the percentage of late orders**


1.  **Input:** any provided promised delivery time and each pair of specified zone and time block
2.   **Return:** the percentage of late order



```python
def Late_Rate_Fun(zone, time_block, promised_time):
  final = np.array([])
  if zone == 'all' and time_block == 'all':
    for z in ZONES:
      for tb in TIME_BLOCKS:
        lst = delivery_times(z, tb, promised_time, seed = 12)
        final = np.append(final,lst)
  elif zone == 'all':
    for z in ZONES:
      lst = delivery_times(z, time_block, promised_time, seed = 12)
      final = np.append(final, lst)
  elif time_block == 'all':
    for tb in TIME_BLOCKS:
      lst = delivery_times(zone, tb, promised_time, seed = 12)
      final = np.append(final, lst)
  else:
    final = delivery_times(zone, time_block, promised_time, seed = 12)
  count = 0
  for time in final:
    if time > promised_time:
      count += 1
  late_rate = count/len(final)*100
  return round(late_rate,2)

```

#### **b. Sort function with descending order of late arrival rate**


1.   **Input:** the current promised delivery time of 45 minutes
2.   **Output:** rank the pairs of zone and time block in the descending order of late arrival rate



```python
def sort_late_rate(time):
  final = []
  for z in ZONES:
      for tb in TIME_BLOCKS:
        late_rate = Late_Rate_Fun(z,tb, time)
        final.append([z, tb, late_rate])
  for i in range(len(final)):
      for j in range(i + 1, len(final)):
          if final[i][2] < final[j][2]:
              temp = final[i]
              final[i] = final[j]
              final[j] = temp
  for txz in final:
    print(txz)
```

#### **c. Function about the average delivery time**


1.  **Input:** any provided promised delivery time and each pair of specified zone and time block
2.   **Return:** the average delivery time

```python
def Ave_Del_Fun(zone, time_block, promised_time):
  final = np.array([])
  if zone == 'all' and time_block == 'all':
    for z in ZONES:
      for tb in TIME_BLOCKS:
        lst = delivery_times(z, tb, promised_time, seed = 12)
        final = np.append(final,lst)
  elif zone == 'all':
    for z in ZONES:
      lst = delivery_times(z, time_block, promised_time, seed = 12)
      final = np.append(final, lst)
  elif time_block == 'all':
    for tb in TIME_BLOCKS:
      lst = delivery_times(zone, tb, promised_time, seed = 12)
      final = np.append(final, lst)
  else:
    final = delivery_times(zone, time_block, promised_time, seed = 12)
  return round(float(final.mean()),2)
```

# **Part II: Exploring the best promise**
#### **a. Function about each late order is costly for Rosas**


1.  **Input:**  

      *  Direct cost due to refund issued to customers
      *  Each late order leads to customer churn: the customer places fewer orders in the future.

2.   **Return:** the total cost per late order

```python
def late_order_cost(cost):
  return cost.get('refund') + cost.get('churn_orders')*cost.get('margin')
```

#### **c. function to choose the best promise**


1.  **Input:** each pair of zone and time block, the list of promises to try and the costs
2.   **Return:** the net profit -- the profit collected from the customers less the late costs

```python
def best_promise_time(zone, time, range_promise, cost):
  final = []
  late_cost = late_order_cost(cost)
  for promise in range_promise:
    late_order = 0
    lst = delivery_times(zone, time, promise, seed = 12)
    profit = cost.get('margin') * len(lst)
    for i in range(len(lst)):
      if lst[i] > promise:
        late_order += 1
    net_profit = profit - late_order * late_cost
    if len(final) == 0:
      final.append(promise)
      final.append(net_profit)
    elif final[1] < net_profit:
      final=[promise, round(net_profit,2)]
  return final[0], final[1]
```

## **Test cases**
**Range:** I chose a 5-minute interval because it provides a reasonable balance between the level of detail and ease of analysis. Since Rosa's current promised delivery time is 45 minutes, I used 45 minutes as a reference point and tested values both below and above it. I first tested 30, 35, 40, and 45 minutes. Although 45 minutes produced the highest net profit within this range, the net profit was still negative. I then extended the range upward and tested 45, 50, 55, 60, and 65 minutes. The net profit reached its highest value at 55 minutes and then decreased, suggesting that the optimal promise was captured within this range. Therefore, I selected 45 to 65 minutes as the final range for testing.

```python
best_promise_time("Far West", "Fri/Sat eve", [30,35,40,45],COSTS)
```
